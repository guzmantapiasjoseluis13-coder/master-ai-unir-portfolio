```python
# ==========================================
# TRABAJO 3: ANÁLISIS DE DATOS CON NUMPY Y PANDAS
# Empresa: RetailNow
# ==========================================

import os
import numpy as np
import pandas as pd

print("Librerías importadas correctamente.")

# Detección de ruta robusta adaptada a la estructura actual
if os.path.exists("workspace/sales.csv"):
    path = "workspace/"
elif os.path.exists("../workspace/sales.csv"):
    path = "../workspace/"
elif os.path.exists("/workspace/sales.csv"):
    path = "/workspace/"
else:
    path = "./workspace/"

print(f"Ruta de datos establecida en: {path}")

# ==========================================
# 1. CARGA Y LIMPIEZA DE DATOS (PANDAS)
# ==========================================
sales_df = pd.read_csv(path + "sales.csv").dropna()
inventories_df = pd.read_csv(path + "inventories.csv").dropna()
satisfaction_df = pd.read_csv(path + "satisfaction.csv").dropna()

print("\n--- Vista previa: Ventas (Limpiado) ---")
display(sales_df.head())

print("\n--- Vista previa: Inventarios (Limpiado) ---")
display(inventories_df.head())

print("\n--- Vista previa: Satisfacción (Limpiado) ---")
display(satisfaction_df.head())


# ==========================================
# 2. ANÁLISIS DE DATOS (PANDAS)
# ==========================================
# Cálculo de ingresos totales por venta (Cantidad * Precio Unitario)
sales_df['Ingreso_Total'] = sales_df['Cantidad_Vendida'] * sales_df['Precio_Unitario']

# Ventas totales agrupadas por tienda
ventas_por_tienda = sales_df.groupby('ID_Tienda')['Ingreso_Total'].sum().reset_index()

print("\n--- Ingresos Totales por Tienda ---")
display(ventas_por_tienda)

print("\n--- Resumen Estadístico de Ventas ---")
display(sales_df.describe())


# ==========================================
# 3. CÁLCULOS ESTADÍSTICOS (NUMPY)
# ==========================================
# Conversión de los ingresos a un array de NumPy para los cálculos
ventas_array = sales_df['Ingreso_Total'].to_numpy()

mediana_ventas = np.median(ventas_array)
desvest_ventas = np.std(ventas_array)
media_ventas = np.mean(ventas_array)

print("\n--- Cálculos Estadísticos (NumPy) ---")
print(f"Ingreso Mediano: {mediana_ventas:.2f}")
print(f"Desviación Estándar de Ingresos: {desvest_ventas:.2f}")
print(f"Ingreso Promedio (Media): {media_ventas:.2f}")


# ==========================================
# 4. SIMULACIÓN DE DATOS (NUMPY)
# ==========================================
# Uso de semilla fija para garantizar reproducibilidad en las proyecciones
np.random.seed(42)

# Simulación de 10 proyecciones de ventas basadas en distribución normal
proyeccion_ventas = np.random.normal(loc=media_ventas, scale=desvest_ventas, size=10)

print("\n--- Simulación de Ventas Futuras (Semilla 42) ---")
for i, sim in enumerate(proyeccion_ventas, 1):
    print(f"Simulación Mes {i}: {sim:.2f}")
```
